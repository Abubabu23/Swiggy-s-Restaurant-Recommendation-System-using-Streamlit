```python
from sklearn.preprocessing import OneHotEncoder
import pickle
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Abuthahir\project\Swiggy Food Recommentation System\data\cleaned_data.csv")
```

```python
cat_featurs = ['city','cuisine']
encoder = OneHotEncoder(sparse_output=False)
enocoder_array = encoder.fit_transform(df[cat_featurs])
```

```python
with open('encoder.pkl','wb')as f:
    pickle.dump(encoder,f)
```

```python
num_features = df[['rating','cost']].values
encoded_data = pd.DataFrame(
    data = np.hstack((num_features,enocoder_array)),
    columns = ['rating','cost'] + list(encoder.get_feature_names_out(cat_featurs))
)


encoded_data.to_csv('encoded_data.csv',index=False)
print("encoded data  saved")
```
